```python
import matplotlib.pyplot as plt

def ic_results_table(results):
    rows = []
    for signal, r in results.items():
        rows.append({
            "signal": signal,
            "mean_ic": r["mean_ic"],
            "ic_tstat": r["ic_tstat"],
            "hit_rate": r["hit_rate"],
            "n_months": r["n_months"],
            "avg_names": r["avg_names"],
        })
    return pd.DataFrame(rows).sort_values("mean_ic", ascending=False)

ic_table = ic_results_table(results)
ic_table
```

```python
def build_rolling_ic_panel(results, window=24):
    frames = []

    for signal, r in results.items():
        ic = r["monthly_ics"].sort_index()
        rolling_mean = ic.rolling(window).mean()
        rolling_std = ic.rolling(window).std()
        rolling_tstat = rolling_mean / (rolling_std / np.sqrt(window))

        frame = pd.DataFrame({
            "date": ic.index,
            "signal": signal,
            "monthly_ic": ic.values,
            "rolling_ic": rolling_mean.values,
            "rolling_tstat": rolling_tstat.values,
        })

        frames.append(frame)

    return pd.concat(frames, ignore_index=True)

rolling_panel = build_rolling_ic_panel(results, window=24)
rolling_panel.head()
```

```python
top_signals = ic_table.head(10)["signal"].tolist()

for signal in top_signals:
    tmp = rolling_panel[rolling_panel["signal"] == signal].copy()

    plt.figure(figsize=(12, 4))
    plt.plot(tmp["date"], tmp["monthly_ic"], alpha=0.25, label="Monthly IC")
    plt.plot(tmp["date"], tmp["rolling_ic"], linewidth=2, label="24M Rolling IC")
    plt.axhline(0, linestyle="--")
    plt.title(signal)
    plt.legend()
    plt.show()
```

```python
for signal in top_signals:
    tmp = rolling_panel[rolling_panel["signal"] == signal].copy()

    plt.figure(figsize=(12, 4))
    plt.plot(tmp["date"], tmp["rolling_tstat"], label="24M Rolling t-stat")
    plt.axhline(0, linestyle="--")
    plt.axhline(2, linestyle="--", label="t=2")
    plt.title(f"Rolling t-stat: {signal}")
    plt.legend()
    plt.show()
```
